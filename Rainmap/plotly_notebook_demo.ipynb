# 在 Jupyter Notebook 中使用 Plotly 进行交互式数据可视化

本 Notebook 演示如何在 Jupyter Notebook 环境下，使用 Plotly 库进行艺术感十足的交互式数据可视化。

## 1. 安装并导入 Plotly 库

如果尚未安装 Plotly，可以使用如下命令进行安装。随后导入所需模块。

```python
# 如未安装 plotly，请先运行：!pip install plotly
import plotly.graph_objs as go
import plotly.io as pio
import numpy as np
```

## 2. 在 Jupyter Notebook 中启用 Plotly 模式

使用 `plotly.offline.init_notebook_mode()` 启用 notebook 模式，确保图表可以在 notebook 内部渲染。

```python
from plotly.offline import init_notebook_mode
init_notebook_mode(connected=True)
```

## 3. 创建基础折线图

用 plotly.graph_objs 绘制一个简单的折线图，并在 notebook 中显示。

```python
# 生成示例数据
t = np.linspace(0, 2 * np.pi, 100)
y = np.sin(t)

trace = go.Scatter(x=t, y=y, mode='lines', name='Sine Wave')
layout = go.Layout(title='基础折线图示例', xaxis=dict(title='t'), yaxis=dict(title='sin(t)'))
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

## 4. 绘制交互式散点图

生成带有悬浮提示和缩放功能的交互式散点图。

```python
# 生成随机散点数据
np.random.seed(42)
x = np.random.randn(100)
y = np.random.randn(100)

scatter = go.Scatter(x=x, y=y, mode='markers', marker=dict(size=12, color=y, colorscale='Viridis', showscale=True),
                     text=[f'点{i}' for i in range(100)], name='随机点')
layout2 = go.Layout(title='交互式散点图', xaxis=dict(title='X'), yaxis=dict(title='Y'))
fig2 = go.Figure(data=[scatter], layout=layout2)
fig2.show()
```

## 5. 自定义图表布局与样式

调整图表的标题、坐标轴标签、颜色等布局和样式参数，提升艺术感。

```python
# 艺术感自定义示例
custom_layout = go.Layout(
    title=dict(text='艺术感折线图', font=dict(size=28, family='Arial Black', color='darkblue')),
    xaxis=dict(title='自定义X轴', showgrid=False, zeroline=False, linecolor='purple', linewidth=2),
    yaxis=dict(title='自定义Y轴', showgrid=True, gridcolor='lightpink', zeroline=False),
    plot_bgcolor='lavender',
    paper_bgcolor='mintcream',
    font=dict(family='Comic Sans MS, cursive', size=16, color='black')
)
fig3 = go.Figure(data=[trace], layout=custom_layout)
fig3.show()
```

## 6. 在 Notebook 中显示 Plotly 图表

使用 `plotly.offline.iplot()` 或 `plotly.io.show()` 在 Jupyter Notebook 中直接显示交互式图表。

```python
# 也可以用 iplot 方式显示
from plotly.offline import iplot
iplot(fig)
iplot(fig2)
iplot(fig3)
```
